```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# loading data
RAW_DIR = Path("../data/raw")

industry_returns = pd.read_parquet(RAW_DIR / "industry_returns.parquet")
ff3_factors = pd.read_parquet(RAW_DIR / "ff3_factors.parquet")
baseline = pd.read_parquet(RAW_DIR / "baseline_momentum.parquet")
residual = pd.read_parquet(RAW_DIR / "residual_momentum.parquet")
```

```python
print("industry_returns: ", industry_returns.shape)
print("ff3_factors: ", ff3_factors.shape)
print("baseline: ", baseline.shape)
print("residual: ", residual.shape)
```

```python
print("Date range:", industry_returns.index.min().date(), "to", industry_returns.index.max().date())
print("Number of months:", len(industry_returns))
```

```python
print("Industries:")
print(industry_returns.columns.tolist())
```

```python
print("Industry returns for Jan 1990:")
print(industry_returns.loc["1990-01-31"].head(10))

print("\nBaseline momentum signal for Jan 1990:")
print(baseline.loc["1990-01-31"].head(10))

print("\nResidual momentum signal for Jan 1990:")
print(residual.loc["1990-01-31"].head(10))
```

```python
# residual signal should be later because it requires the 36 month residualization window + 12 month momentum window
print("First date with valid baseline signal:")
print(baseline.dropna(how="all").index[0])

print("\nFirst date with valid residual signal:")
print(residual.dropna(how="all").index[0])
```

```python
# comparing baseline vs. residual for a given date where both signals exist
date = "1990-01-31"

comparison = pd.DataFrame({
    "baseline": baseline.loc[date],
    "residual": residual.loc[date],
})
comparison["diff"] = comparison["residual"] - comparison["baseline"]
print(comparison.sort_values("baseline", ascending=False).head(15))
```
